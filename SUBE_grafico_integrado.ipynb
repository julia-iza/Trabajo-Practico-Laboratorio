# SUBE: gráfico integrado

El gráfico reúne indicadores históricos de uso y porcentajes de descuento. Los indicadores históricos tienen denominadores distintos y no constituyen una serie comparable; los descuentos muestran reglas del esquema, no la proporción de viajes que efectivamente los utilizó.

## Código Python para reproducir el gráfico

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Datos
labels = [
    "Viajes con tarifa plena (2015)",
    "Usuarios que pagaban tarifa plena (2018)",
    "Personas con subsidio directo (2018)",
    "Personas que accedían a Red SUBE (2018)",
    "Descuento Red SUBE: 2.º viaje",
    "Descuento Tarifa Social",
    "Descuento Red SUBE: 3.º viaje y siguientes"
]

values = [90, 40, 60, 43, 50, 55, 75]
colors = ["#4472C4"] * 4 + ["#ED7D31"] * 3

# Crear gráfico
fig, ax = plt.subplots(figsize=(12, 7.2))
y = range(len(labels))
bars = ax.barh(y, values, color=colors, height=0.62)

ax.set_yticks(y)
ax.set_yticklabels(labels, fontsize=10)
ax.invert_yaxis()
ax.set_xlim(0, 100)
ax.set_xlabel("Porcentaje (%)", fontsize=11)
ax.set_title(
    "SUBE: uso del sistema y niveles de descuento",
    fontsize=16, pad=18, weight="bold"
)
ax.xaxis.grid(True, linestyle="--", alpha=0.35)
ax.set_axisbelow(True)

# Porcentajes al final de las barras
for bar, value in zip(bars, values):
    ax.text(
        value + 1.2,
        bar.get_y() + bar.get_height() / 2,
        f"{value}%",
        va="center", fontsize=10, weight="bold"
    )

# Leyenda
legend_items = [
    Patch(facecolor="#4472C4",
          label="Indicadores históricos de uso (2015-2018)"),
    Patch(facecolor="#ED7D31",
          label="Porcentajes de descuento del esquema")
]
ax.legend(handles=legend_items, loc="lower right",
          frameon=False, fontsize=9)

# Aclaración metodológica
fig.text(
    0.12, 0.015,
    "Importante: los indicadores históricos tienen denominadores "
    "distintos (viajes, usuarios o personas) y no forman una serie "
    "comparable. Los descuentos representan reglas de beneficio, "
    "no la proporción de viajes que efectivamente los utilizó. "
    "Red SUBE: 50% en el segundo viaje y 75% desde el tercero "
    "dentro de la ventana de integración; Tarifa Social: 55%.",
    ha="left", va="bottom", fontsize=8.5, wrap=True
)

plt.subplots_adjust(left=0.34, right=0.94, top=0.88, bottom=0.22)

# Guardar y mostrar
plt.savefig("grafico_integrado_SUBE.png",
            dpi=200, bbox_inches="tight")
plt.show()
```
